```python
from transformers import AutoModelForSequenceClassification

model = AutoModelForSequenceClassification.from_pretrained(
    "FacebookAI/xlm-roberta-base",
    num_labels=2
)

model = model.to("cuda")

print("XLM-R model loaded successfully!")
print("Model device:", next(model.parameters()).device)
```

```python
# Make sure labels are numeric
train_df["label_num"] = train_df["label"].map({"real": 0, "fake": 1})
val_df["label_num"] = val_df["label"].map({"real": 0, "fake": 1})
test_df["label_num"] = test_df["label"].map({"real": 0, "fake": 1})

print("Train labels:")
print(train_df["label_num"].value_counts())

print("\nValidation labels:")
print(val_df["label_num"].value_counts())

print("\nTest labels:")
print(test_df["label_num"].value_counts())
```

```python
from datasets import Dataset

train_data = Dataset.from_pandas(
    train_df[["text", "label_num"]],
    preserve_index=False
)

val_data = Dataset.from_pandas(
    val_df[["text", "label_num"]],
    preserve_index=False
)

test_data = Dataset.from_pandas(
    test_df[["text", "label_num"]],
    preserve_index=False
)

print("Train:", train_data)
print("Validation:", val_data)
print("Test:", test_data)
```

```python
def tokenize_data(batch):
    return tokenizer(
        batch["text"],
        truncation=True,
        padding="max_length",
        max_length=128
    )

train_tokenized = train_data.map(
    tokenize_data,
    batched=True,
    batch_size=1000
)

val_tokenized = val_data.map(
    tokenize_data,
    batched=True,
    batch_size=1000
)

test_tokenized = test_data.map(
    tokenize_data,
    batched=True,
    batch_size=1000
)

print("Tokenization completed!")
print("Train:", train_tokenized)
print("Validation:", val_tokenized)
print("Test:", test_tokenized)
```

```python
# Rename label column for the model
train_tokenized = train_tokenized.rename_column("label_num", "labels")
val_tokenized = val_tokenized.rename_column("label_num", "labels")
test_tokenized = test_tokenized.rename_column("label_num", "labels")

# Remove original text column
train_tokenized = train_tokenized.remove_columns(["text"])
val_tokenized = val_tokenized.remove_columns(["text"])
test_tokenized = test_tokenized.remove_columns(["text"])

print("Train columns:", train_tokenized.column_names)
print("Validation columns:", val_tokenized.column_names)
print("Test columns:", test_tokenized.column_names)
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="/content/drive/MyDrive/Fake_News_Project/xlm_r_results",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=2,
    weight_decay=0.01,
    logging_steps=500,
    load_best_model_at_end=True,
    metric_for_best_model="accuracy",
    report_to="none"
)

print("Training configuration created successfully!")
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

def compute_metrics(eval_pred):
    logits, labels = eval_pred

    predictions = np.argmax(logits, axis=-1)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="binary",
        zero_division=0
    )

    accuracy = accuracy_score(labels, predictions)

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }

print("Evaluation metrics ready!")
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_tokenized,
    eval_dataset=val_tokenized,
    compute_metrics=compute_metrics
)

print("Trainer created successfully!")
```

```python
train_result = trainer.train()
```

```python
def extract_image_features(urls, model, transform):
    features = []
    valid_indices = []

    for i, url in enumerate(urls):
        try:
            response = requests.get(url, timeout=10)

            if response.status_code != 200:
                continue

            image = Image.open(
                BytesIO(response.content)
            ).convert("RGB")

            image_tensor = transform(image).unsqueeze(0).to("cuda")

            with torch.no_grad():
                feature = model(image_tensor)

            features.append(feature.cpu().numpy().flatten())
            valid_indices.append(i)

        except Exception:
            continue

    return np.array(features), valid_indices

print("Image feature extraction function ready!")
```

```python
# FAST IMAGE FEATURE EXTRACTION
# Parallel downloading + T4 GPU batching + checkpoints

from concurrent.futures import ThreadPoolExecutor
from PIL import Image
from io import BytesIO
import requests
import numpy as np
import os
import time
import torch

project_path = "/content/drive/MyDrive/Fake_News_Project"

# Settings
download_workers = 16
batch_size = 500
checkpoint_every = 5000

all_features = []
all_indices = []

total = len(multimodal_df)

def download_image_fast(url):
    try:
        response = requests.get(
            url,
            timeout=8,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        if response.status_code == 200:
            return Image.open(
                BytesIO(response.content)
            ).convert("RGB")

    except Exception:
        return None

    return None


start_time = time.time()

for start in range(0, total, batch_size):

    end = min(start + batch_size, total)

    batch_urls = multimodal_df["image"].iloc[start:end].tolist()

    # Parallel download
    with ThreadPoolExecutor(max_workers=download_workers) as executor:
        images = list(executor.map(download_image_fast, batch_urls))

    # Keep valid images
    valid_images = []
    valid_indices = []

    for j, img in enumerate(images):
        if img is not None:
            valid_images.append(img)
            valid_indices.append(start + j)

    # GPU batch processing
    if len(valid_images) > 0:

        tensors = torch.stack([
            image_transform(img)
            for img in valid_images
        ]).to(device)

        with torch.no_grad():
            features = image_model(tensors)

        features = features.cpu().numpy()

        all_features.append(features)
        all_indices.extend(valid_indices)

    processed = end

    if processed % checkpoint_every == 0 or processed == total:

        temp_features = np.vstack(all_features)

        np.save(
            f"{project_path}/image_features_checkpoint.npy",
            temp_features
        )

        np.save(
            f"{project_path}/image_indices_checkpoint.npy",
            np.array(all_indices)
        )

        print(
            f"Checkpoint saved: {processed}/{total} | "
            f"Valid images: {len(all_indices)}"
        )

elapsed = time.time() - start_time

print("\nIMAGE FEATURE EXTRACTION COMPLETE!")
print("Total processed:", total)
print("Valid images:", len(all_indices))
print("Failed images:", total - len(all_indices))
print("Feature shape:", np.vstack(all_features).shape)
print("Time taken:", round(elapsed / 60, 2), "minutes")
```
